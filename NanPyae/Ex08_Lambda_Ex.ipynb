Q1). Write a lambda function that takes an integer 'year' and returns 'age'.

```python
from datetime import datetime

current_year = datetime.now().year

your_age = lambda birth_year, current_year: current_year - birth_year

birth_year = int(input("Enter your birth year:"))
your_age(birth_year, current_year)
```

Q2) Write a lambda function that takes a float "cost" and returns a float "price" by adding 10% to the cost.

```python
cost = float(input("Enter the cost:"))

calculate_price = lambda cost: cost + cost*0.1

print(f"The price is: {calculate_price(cost)}")
```

Q3) Write a lambda function that takes two strings 'name' , 'domain' and returns an email address (string).

```python
#accept the name for email and remove space between words
name = input("Enter name for email:").replace(" ","")
domain = input("Enter domain for email:")

email_address = lambda name, domain: name + '@' + domain
print(f"Your email is: {email_address(name, domain)}")
```
